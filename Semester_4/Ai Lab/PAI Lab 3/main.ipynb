```python
# Applying rules and making Main menu
def main_menu(x, y):
    print(f"\nCurrent State: x = {x}, y = {y}")
    print("Choose an action:")
    for i, rule in enumerate(rules.keys(), 1):
        print(f"{i}. {rule}")
    print("0. Exit")

    choice = int(input("Enter your choice: "))
    if choice == 0:
        return None
    elif 1 <= choice <= len(rules):
        rule_name = list(rules.keys())[choice - 1]
        return rules[rule_name](x, y)
    else:
        print("Invalid choice. Please try again.")
        return (x, y)
# Main loop
while True:
    result = main_menu(x, y)
    if result is None:
        print("Exiting the program.")
        break
    else:
        x, y = result
```

```python
from collections import deque

# Jug capacities
max_x = 3  # Jug B
max_y = 4  # Jug A

# Goal: 2 liters in either jug
goal = 2

# Rules using lambda functions
rules = {
    "Fill x": (lambda x, y: (max_x, y)),
    "Fill y": (lambda x, y: (x, max_y)),
    "Empty x": (lambda x, y: (0, y)),
    "Empty y": (lambda x, y: (x, 0)),
    "Pour x to y": (lambda x, y: (x - min(x, max_y - y), y + min(x, max_y - y))),
    "Pour y to x": (lambda x, y: (x + min(y, max_x - x), y - min(y, max_x - x))),
}

# Breadth-First Search to find solution path
def bfs():
    visited = set()
    queue = deque()
    start = (0, 0)
    queue.append((start, []))  # (state, path)

    while queue:
        (x, y), path = queue.popleft()
        
        # Goal check
        if x == goal or y == goal:
            return path + [((x, y), "Goal reached")]

        if (x, y) in visited:
            continue

        visited.add((x, y))

        # Generate next states using map
        next_states = list(map(
            lambda rule: (rules[rule](x, y), rule),
            rules
        ))

        # Filter unvisited states
        next_states = list(filter(
            lambda item: item[0] not in visited,
            next_states
        ))

        # Enqueue new states with updated paths
        for (new_state, action) in next_states:
            queue.append((new_state, path + [((x, y), action)]))

    return None  # No solution found

# Run the search
solution = bfs()

# Print the solution path
if solution:
    print("\nSolution steps:")
    for (state, action) in solution:
        print(f"{action} => x = {state[0]}, y = {state[1]}")
else:
    print("No solution found.")
```
